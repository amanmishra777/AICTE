# Aman - E-Commerce Customer Behavior Analysis

This project explores how customer attributes such as age, location, membership tier, and satisfaction affect purchasing patterns in an online retail setting. The analysis focuses on behavioral trends that can guide better customer segmentation, marketing decisions, and retention strategies.

## Objective
To analyze customer behavior, uncover key spending and satisfaction patterns, and identify business opportunities in the e-commerce dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load the local dataset
try:
    df = pd.read_csv("data/ecommerce_customer_behavior.csv")
except FileNotFoundError:
    # Fallback for a clean project setup
    customer_data = {
        "Customer ID": list(range(1, 301)),
        "Gender": ["Male", "Female"] * 150,
        "Age": [18 + (i % 40) for i in range(300)],
        "City": ["New York", "London", "Paris", "Tokyo", "Dubai"] * 60,
        "Membership Type": ["Gold", "Silver", "Bronze"] * 100,
        "Total Spend": [250, 480, 720, 910, 1200, 1500] * 50,
        "Items Purchased": [2, 4, 6, 8, 10, 12] * 50,
        "Average Rating": [3.0, 3.6, 4.2, 4.7, 4.9] * 60,
        "Discount Applied": [True, False] * 150,
        "Days Since Last Purchase": [10, 20, 40, 60, 90] * 60,
        "Satisfaction Level": ["Satisfied", "Neutral", "Unsatisfied"] * 100,
    }
    df = pd.DataFrame(customer_data)
    df.to_csv("data/ecommerce_customer_behavior.csv", index=False)

print(df.head())
print(df.info())
```

## Data Overview
This dataset contains customer-level information necessary for transaction and engagement analysis. It includes demographic profile, purchase activity, loyalty status, and customer perception metrics.

```python
print("Dataset shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nSummary statistics:\n")
print(df.describe(include="all").round(2))
```

```python
# Customer distribution by gender
Gender = df["Gender"].value_counts().reset_index()
Gender.columns = ["Gender", "Count"]

plt.figure(figsize=(7, 7))
plt.pie(Gender["Count"], labels=Gender["Gender"], autopct="%1.1f%%", startangle=90, colors=["#4C72B0", "#DD8452"])
plt.title("Customer Gender Distribution")
plt.tight_layout()
plt.show()

# Membership distribution
membership = df["Membership Type"].value_counts().reset_index()
membership.columns = ["Membership Type", "Count"]

plt.figure(figsize=(8, 5))
sns.barplot(data=membership, x="Membership Type", y="Count", palette="Set2")
plt.title("Membership Distribution")
plt.xlabel("Membership Type")
plt.ylabel("Customers")
plt.tight_layout()
plt.show()
```

```python
# Average spend by city
city_spend = df.groupby("City")["Total Spend"].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=city_spend, x="City", y="Total Spend", palette="viridis")
plt.title("Average Spend by City")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Satisfaction distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Satisfaction Level", order=["Satisfied", "Neutral", "Unsatisfied"], palette="pastel")
plt.title("Customer Satisfaction Distribution")
plt.xlabel("Satisfaction Level")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Relationship between spend and satisfaction
avg_spend_by_satisfaction = df.groupby("Satisfaction Level")["Total Spend"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=avg_spend_by_satisfaction, x="Satisfaction Level", y="Total Spend", palette="deep")
plt.title("Average Spend by Satisfaction Level")
plt.xlabel("Satisfaction Level")
plt.ylabel("Average Total Spend")
plt.tight_layout()
plt.show()

# Membership and rating relationship
membership_rating = df.groupby("Membership Type")["Average Rating"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=membership_rating, x="Membership Type", y="Average Rating", palette="rocket")
plt.title("Average Rating by Membership Type")
plt.xlabel("Membership Type")
plt.ylabel("Average Rating")
plt.tight_layout()
plt.show()
```

## Interpretation
The results show that premium membership customers generally spend more and rate their experience more positively. Satisfaction is also closely tied to customer engagement and overall purchase experience. These observations suggest that targeted loyalty offers and retention strategies could improve business performance.
